```python
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Generate a 1x10 vector of random numbers between 0 and 1
uniform_data = np.random.rand(1, 10)
print(f"Uniform Vector (1x10):\n{uniform_data}\n")

# Step 2: Plot a histogram of the vector
plt.figure()
plt.hist(uniform_data.flatten(), bins=5, color="skyblue", edgecolor="black")
plt.title("Histogram of Uniform Random Numbers (1x10)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Generate 500 random numbers from a standard normal distribution (mean=0, std=1)
normal_data = np.random.normal(loc=0, scale=1, size=500)

# Step 2: Plot a histogram to visualize the distribution
plt.figure()
plt.hist(
    normal_data,
    bins=30,
    color="lightgreen",
    edgecolor="black",
    density=True,
)
plt.title("Histogram of Standard Normal Distribution (N=500)")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

```python
import numpy as np

# Part A: Create a 3x3 matrix WITH a fixed seed (np.random.seed(10))
np.random.seed(10)
fixed_seed_matrix = np.random.randint(1, 51, size=(3, 3))
print("3x3 Matrix WITH fixed seed (np.random.seed(10)):")
print(f"{fixed_seed_matrix}\n")

# Part B: Create a 3x3 matrix WITHOUT a fixed seed
np.random.seed(None)
no_seed_matrix = np.random.randint(1, 51, size=(3, 3))
print("3x3 Matrix WITHOUT fixed seed:")
print(f"{no_seed_matrix}\n")

# Part C: Explanation of differences
print(
    "Explanation:\n"
    "- Fixed Seed: Fixing the seed sets the random number generator to a specific initial state, "
    "making the generated values 100% reproducible every time you run the code.\n"
    "- Without Fixed Seed: The generator uses entropy from system time, causing the matrix values "
    "to change continuously on every execution."
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# Step 1: Define the dataset (observed sample)
data = np.array([5.2, 5.4, 5.1, 5.6, 5.3, 5.8])

# Step 2: Define the population mean (hypothesized mean)
population_mean = 5.5

# Step 3: Perform the one-sample t-test
t_statistic, p_value = stats.ttest_1samp(data, population_mean)

# Step 4: Display the results
print(f"Sample Mean: {np.mean(data):.2f}")
print(f"Hypothesized Population Mean: {population_mean:.2f}")
print(f"t-Statistic: {t_statistic:.4f}")
print(f"Degrees of Freedom: {len(data) - 1}")
print(f"p-Value: {p_value:.4f}")

# Step 5: Conclusion / Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print(
        f"Reject the null hypothesis. The average score is significantly different from {population_mean:.2f}."
    )
else:
    print(
        f"Fail to reject the null hypothesis. The average score is not significantly different from {population_mean:.2f}."
    )

# Step 6: Visualize the data
plt.figure()
plt.bar(1, np.mean(data), color="skyblue", label="Sample Mean")
plt.errorbar(
    1, np.mean(data), yerr=np.std(data, ddof=1), fmt="k", linewidth=1.5
)
plt.axhline(
    y=population_mean,
    color="r",
    linestyle="--",
    linewidth=1.5,
    label="Hypothesized Mean",
)
plt.title("Sample Mean vs Hypothesized Mean")
plt.xlim(0.5, 1.5)
plt.xlabel("Sample")
plt.ylabel("Mean Value")
plt.legend()
plt.grid(True)
plt.show()
```
